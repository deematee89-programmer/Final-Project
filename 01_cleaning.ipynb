# **5. Handle Missing Values**

#<><><><><><><><><><><><><><><><><><><><><><><><><><><><><><><><><><><><><>

#

*   Here i will handle the missing values , i fill each coloumn in a  suitable way based on its type:

1.   **Car col**:filled with the median because it is a numeric column and the median is less affected by outliers.
2.   **Building Area col**:filled with the median to keep reasonable values without removing many rows.
3. **Year Built col**:filled with the median because it is numeric and the median gives a stable value.
4. **Council Area col**: fill with "Unknown" because its categorical coloumn





####===============================================================================================

```python
def handle_missing_values(df):
  df["Car"]=df["Car"].fillna(df["Car"].median())
  df["BuldingArea"]=df["BuildingArea"].fillna(df["BuildingArea"].median())
  df["YearBuilt"]=df["YearBuilt"].fillna(df["YearBuilt"].median())
  df["CouncilArea"]=df["CouncilArea"].fillna("Unknown")
  return df

missing_values=handle_missing_values(df)
```

####-------------------------------------------------------------------------------------------------------------------------

## -**Fix coloumns types**

####-------------------------------------------------------------------------------------------------------------------------

### - i have wrong Data types on my data coloumns so here i will fix them using "astype()":
* "datetime" for **Date** coloumn and i use **dayfirst** so the program can understand its a date.
*  "string" for **postcode** coloumn.
*  "int" for **Bedroom , Bathroom , Car** coloumns.



####-------------------------------------------------------------------------------------------------------------------------

```python
def fix_coloumns_types(df):
  df["Date"]=pd.to_datetime(df["Date"],dayfirst=True)
  df["Postcode"]=df["Postcode"].astype(str)
  df["Bedroom2"]=df["Bedroom2"].astype(int)
  df["Bathroom"]=df["Bathroom"].astype(int)
  df["Car"]=df["Car"].astype(int)
  return df

fix_col=fix_coloumns_types(df)
```

# -Exploring Outliers using Seaborn and Matplotlib

- I used the **boxplot from Seaborn** because it is one of the best visualizations for detecting outliers. It clearly shows the distribution of the data and highlights any extreme values.

- I used **plt.tight_layout()** to ensure that the plot elements do not overlap and the visualization appears clean and well-organized.

- I experimented with both the **IQR method** and the **Capping method** to handle outliers. After comparing the results, I found that capping helps reduce the effect of extreme values without removing data points from the dataset.


####===============================================================================================

####-------------------------------------------------------------------------------------------------------------------------

## *plot before capping*

####-------------------------------------------------------------------------------------------------------------------------

```python
sns.boxplot(df["Price"])
plt.tight_layout()
```

####------------------------------------------------------------------------------------------------------------------------

## *cap outliers*

####-------------------------------------------------------------------------------------------------------------------------

```python
def cap_outliers(df,price):
  upper=df["Price"].quantile(0.99)
  df["Price"]=df["Price"].clip(upper=upper)
  return df

capping=cap_outliers(df,"Price")
```
